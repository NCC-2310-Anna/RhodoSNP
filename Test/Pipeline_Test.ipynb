# Testing the Installation with a small dataset (pDx from _R. sphaeroides_)

Creating a working directory named `./Data`.
This is the directory where all the magic take place. Then prepare your working directory using the following script:

```bash
# Enter your project dir
# mkdir Data (optional if using not the cloned Github)
# Setup Data
pwd # This should be something like "~/GitHub/RhodoSNP/Test"
cd ./Data
# bash ../working_directory_setup.sh -d . #if setup is not already done
```

Let's take a look into the structure of our working directory. The `working_directory_setup.sh` provides us a Readme.md which shows the structure. Or we can plot the file tree...

```bash
tree .
```

The input directory is already filled with a reference sequence (`Rsph_orig.fa`), a genome annotation (`Rsph.gff3`) and simulated illumina reads (`reads_R1.fastq` and `reads_R2.fastq`). The output directory will be filled in the following steps by RhodoSNP...

```bash
conda activate RhodoSNP
bash ../../Calling.sh -d . -r Rsph_orig.fa -t 20 -p 1 -g Rsph.gff3
```

The interesting outputs are listed in `Data/Output/ProteinSequences/Predictions`: the `reads_annotation.tsv` lists all variants in coding regions and their known location. The `reads_protein.fasta` lists all protein sequences with a predicted amino acid change.

The pipeline runs correctly on your system if the output looks like this:

```bash
head ./Output/ProteinSequences/Predictions/reads_annotation.tsv
```
